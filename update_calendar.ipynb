```python
from ics import Calendar, Event
import os
import sqlite3
from tmdb_info import get_tmdb_movie_info, get_tmdb_seasons
```

```python
working_path = os.pardir
working_path+="\\{}"
```

```python
media_organize_path = working_path.format("media_organize.db")
```

```python
media_organize_path
```

```python
conn_media_organize = sqlite3.connect(media_organize_path)
```

```python
results = conn_media_organize.execute("select tmdb_id from tbl_imdb_list where bool_movie = 1 order by date_added desc limit 50")
rows = results.fetchall()
l_movie_ids = rows

l_movie_ids = [m[0] for m in l_movie_ids]

results = conn_media_organize.execute("select tmdb_id from tbl_imdb_list where bool_movie is null order by date_added desc limit 50")
rows = results.fetchall()
l_tv_ids = rows

l_tv_ids = [t[0] for t in l_tv_ids]

conn_media_organize.close()
```

```python
calendar = Calendar()

for m in l_movie_ids:
    movie_info = get_tmdb_movie_info(m)
    movie_info = {"title":movie_info.get("title"), "release_date":movie_info.get("release_date")}

    event = Event()

    if movie_info["title"]:
        try:
            event.name = movie_info["title"]
            event.begin = movie_info["release_date"]
            event.make_all_day()        

            calendar.events.add(event)
        except:
            pass
```

```python
with open("move_cal.ics", "w") as f:
    f.writelines(calendar.serialize_iter())
```
